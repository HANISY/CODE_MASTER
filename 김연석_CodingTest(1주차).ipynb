```python
# [중급 문제.ver1] !! 문제: reverse의 문제, join에서 뛰어쓰기가 안 나온다. / 해결법: reverse 부분에서 반복문을 하고 join 에서 스페이스바를 입력한다.

# 입력을 받는다.
string = input("문자를 입력해주세요: ")

# split()으로 단어를 쪼갠다.
word = string.split()

# 슬라이싱으로 reverse 한다.
reverse_word = word[::-1]

# join을 한다.
result = ''.join(reverse_word)

print(result)
```

```python
# [중급 문제.ver2] !! 문제: #1의 리스트 변수 덮어쓰기, #2의 append는 리스트 객체에서만 가능 / 
#                   해결법: #1: 리스트 지정    #2: 새로운 변수를 선언   #3: 앞에 선언한 변수와 ()안에 넣기

# 입력을 받는다.
string = input("문자를 입력해주세요: ")

# split()으로 단어를 쪼갠다.
word = string.split()

# 리스트문 이용과 append 사용, 슬라이싱으로 reverse 한다.
reverse_word = []     #1
for i in word:
    reverse_word = i[::-1]   #2
    reverse_word.append(reverse_word)   #3

# join을 한다.
result = ' '.join(reverse_word)

print(result)
```

```python
# [중급 문제.ver3] -- 출력 완료

# 입력을 받는다.
string = input("문자를 입력해주세요: ")

# split()으로 단어를 쪼갠다.
word = string.split()

# 리스트문 이용과 append 사용, 슬라이싱으로 reverse 한다.
reverse_word = []
for i in word:
    reverse_word_string = i[::-1]
    reverse_word.append(reverse_word_string)

# join을 한다.
result = ' '.join(reverse_word)

print(result)
```

```python
# [고급 문제.ver1] !! 문제점: for 루프가 끝난 직후, 남아있는 current_string과 count를 마지막으로 한 번 더 result로 불러야한다.

# 입력을 받는다.
string = input("문자를 입력해주세요: ")

# 결과 문자열, 현재 문자, 카운트가 필요하다.
result = ""
current_string = string[0]
count = 1

for  i in range(1, len(string)):
    if string[i] == current_string:
        count += 1
    else:
        result += current_string + str(count)



print(result)
```
