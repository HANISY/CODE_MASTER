```python
# 초급 문제: 문자열 대소문자 변환기
# 문자열을 입력받아, 소문자는 대문자로, 대문자는 소문자로 바꿔 출력하시오.
# 공백이나 숫자는 그대로 둔다.
def deso(input_string):
    result=[]
    for char in input_string:
        if 'a'<=char<='z':
            result.append(char.upper())
        elif 'A'<=char<='Z':
            result.append(char.lower())
        else:
            result.append(char)
    return "".join(result)
input="PyThOn"
output=deso(input)
print(f"입력: {input}")
print(f"출력: {output}")
```

```python
# 중급 문제: 단어별 뒤집기
# 문장에서 각 단어의 철자 순서를 뒤집되, 단어의 순서는 그대로 유지하시오.
# 공백은 단어 사이 한 칸으로 가정한다.
def reverse(sentence):
    word=sentence.split(' ')
    reversed_word=[]
    for w in word:
        reversed_w=w[::-1]
        reversed_word.append(reversed_w)
    return " ".join(reversed_word)
input="python is fun"
output=reverse(input)
print(f"입력: {input}")
print(f"출력: {output}")
```

```python
# 고급 문제: 압축 문자열(Run-Length Encoding)
# 문자열이 주어졌을 때, 같은 문자가 연속으로 반복되면 반복 횟수를 숫자로 표시해 압축하시오.
# 예: aaabcc -> a3b2c1
def rle(input_string):
    if not input_string:
        return ""
    encoding_string=[]
    current_char=input_string[0]
    count=1
    for i in range(1,len(input_string)):
        if input_string[i]==current_char:
            count+=1
        else:
            encoding_string.append(current_char)
            encoding_string.append(str(count))
            current_char=input_string[i]
            count=1
    encoding_string.append(current_char)
    encoding_string.append(str(count))
    return "".join(encoding_string)
input="aaabbc"
output=rle(input)
print(f"입력: {input}")
print(f"출력: {output}")
```
